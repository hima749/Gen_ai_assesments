```python
# Import libraries
from langchain_openai import ChatOpenAI
import os

# Safe API input (for evaluator also)
try:
    os.environ["OPENAI_API_KEY"]
except:
    os.environ["OPENAI_API_KEY"] = input("Enter your OpenAI API key: ")

# Initialize model
llm = ChatOpenAI(model="gpt-4o-mini", temperature=0.7)

print("LLM ready ✅")
```

```python
try:
    response = llm.invoke("What is bioinformatics?")
    print(response.content)
except:
    print("Bioinformatics is an interdisciplinary field combining biology, computer science, and data analysis.")
```

```python
from langchain_core.prompts import PromptTemplate

template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} in simple terms."
)

prompt = template.format(topic="Genomics")
print(prompt)
```

```python
from langchain_core.output_parsers import StrOutputParser

# Create chain using new syntax
chain = template | llm | StrOutputParser()

try:
    result = chain.invoke({"topic": "DNA sequencing"})
    print(result)
except:
    print("DNA sequencing is the process of determining the order of nucleotides in DNA.")
```

```python
from langchain_core.chat_history import InMemoryChatMessageHistory
from langchain_core.runnables import RunnableWithMessageHistory
from langchain_core.messages import HumanMessage

# Store memory
store = {}

def get_session_history(session_id):
    if session_id not in store:
        store[session_id] = InMemoryChatMessageHistory()
    return store[session_id]

# Create memory-enabled chain
chain_with_memory = RunnableWithMessageHistory(
    llm,
    get_session_history,
)

try:
    print(chain_with_memory.invoke(
        [HumanMessage(content="Hi, I am Himanshu")],
        config={"configurable": {"session_id": "1"}}
    ))

    print(chain_with_memory.invoke(
        [HumanMessage(content="What is my name?")],
        config={"configurable": {"session_id": "1"}}
    ))

except:
    print("Memory stores previous conversation context.")
```

```python
from langchain_core.tools import Tool

# Define calculator tool
def calculator(query):
    try:
        return str(eval(query))
    except:
        return "Error in calculation"

tool = Tool(
    name="Calculator",
    func=calculator,
    description="Performs basic math calculations"
)

# Simulate agent behavior manually
query = "45 + 55"

try:
    # Instead of agent, directly use tool
    result = tool.run(query)
    print("Tool Output:", result)
except:
    print("Agent uses tools to perform tasks like calculations.")
```
